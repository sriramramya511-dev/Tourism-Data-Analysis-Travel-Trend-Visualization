```python
import pandas as pd
import matplotlib.pyplot as plt
# Load the cleaned tourism data from Excel
df = pd.read_excel("Original_and_Cleaned_Tourism_Data-1.xlsx",
sheet_name="Cleaned Data")
```

```python
yearly = df.groupby("Year", as_index=False)[
["International_Arrivals", "Tourism_Receipts_USD",
"Tourism_Expenditures_USD"]
].sum()
plt.figure(figsize=(9, 4.5))
plt.plot(yearly["Year"], yearly["International_Arrivals"] / 1_000_000,
marker="o")
plt.title("Total International Tourist Arrivals by Year")
plt.xlabel("Year")
plt.ylabel("Arrivals (millions)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4.5))
plt.plot(yearly["Year"], yearly["Tourism_Receipts_USD"] / 1e12,
marker="o", label="Tourism receipts")
plt.plot(yearly["Year"], yearly["Tourism_Expenditures_USD"] / 1e12,
marker="s", label="Tourism expenditures")
plt.title("Tourism Receipts vs Expenditures by Year")
plt.xlabel("Year")
plt.ylabel("Amount (trillion USD)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
regional = df.groupby("Region", as_index=False)[
"International_Arrivals"
].sum().sort_values("International_Arrivals", ascending=True)
plt.figure(figsize=(9, 5))
plt.barh(regional["Region"], regional["International_Arrivals"] / 1e6)
plt.title("Cumulative International Arrivals by Region (2015–2025)")
plt.xlabel("Arrivals across years (millions)")
plt.tight_layout()
plt.show()
```
